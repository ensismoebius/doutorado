```python
import numpy as np
```

```python
def framesignal(x, frame_len, hop):
    N = frame_len
    L = len(x)
    num_frames = 1 + (L - N) // hop if L >= N else 1
    frames = np.zeros((num_frames, N))
    for i in range(num_frames):
        start = i * hop
        frames[i] = x[start:start+N]
    window = 0.5 * (1 - np.cos(2*np.pi*np.arange(N)/(N-1)))
    frames *= window
    return frames
```
